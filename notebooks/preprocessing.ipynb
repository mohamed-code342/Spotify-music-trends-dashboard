## 2-) Data Clean

```python
df=df.drop(columns=['Unnamed: 0','track_id'])
df
```

```python
df.isna().sum()
```

```python
df=df.dropna()
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df = df[(df['popularity'] >= 0) & (df['popularity'] <= 100)]
```

```python
df['artists'] = df['artists'].str.strip().str.title()
```

```python
df['track_genre'] = df['track_genre'].str.strip().str.lower()
```

```python
df['artists'] = df['artists'].str.split(';')
df = df.explode('artists')
```

```python
df.head()
```

## 3-) Feature Engineering

```python
df['duration_min'] = df['duration_ms'] / (1000 * 60)
```

```python
df = df[df['duration_min'] < 15]
```

```python
def popularity_level(x):
    if x < 30:
        return 'Low'
    elif x < 70:
        return 'Medium'
    else:
        return 'High'

df['popularity_level'] = df['popularity'].apply(popularity_level)
```

```python
df['duration_category'] = pd.cut(
    df['duration_min'],
    bins=[0,2,4,6,20],
    labels=['Very Short','Short','Medium','Long']
)
df['duration_category']
```

```python
df['tempo_category'] = pd.cut(
    df['tempo'],
    bins=[0,90,120,150,250],
    labels=['Slow','Moderate','Fast','Very Fast']
)
df['tempo_category']
```

```python
df['engagement_score'] = (
    df['energy'] +
    df['danceability'] +
    df['valence']
) / 3
df['engagement_score']
```

```python
df['loudness_level'] = pd.cut(
    df['loudness'],
    bins=4,
    labels=['Low','Medium','High','Very High']
)
df['loudness_level']
```

```python
df['explicit_label'] = df['explicit'].map({
    True: 'Explicit',
    False: 'Non-Explicit'
})
df['explicit_label']
```

```python
df['mood_score'] = (df['valence'] + df['energy']) / 2
df['mood_score']
```

```python
artist_counts = df['artists'].value_counts()
df['artist_song_count'] = df['artists'].map(artist_counts)
df['artist_song_count']
```

```python
genre_rank = df.groupby('track_genre')['popularity'].mean().rank(ascending=False)
df['genre_rank'] = df['track_genre'].map(genre_rank)
df['genre_rank']
```
